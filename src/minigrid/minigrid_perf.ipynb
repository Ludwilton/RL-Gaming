# MiniGrid RL

## Setup

```python
from pathlib import Path

from feature_extractor import MinigridFeaturesExtractor
from levels import LevelOne, LevelTwo
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env
from stable_baselines3.common.vec_env import DummyVecEnv, SubprocVecEnv

from minigrid.wrappers import ImgObsWrapper

SEED = 1123
device = "cpu"  # PPO is meant to run on CPU
policy_kwargs = {
    "features_extractor_class": MinigridFeaturesExtractor,
    "features_extractor_kwargs": {"features_dim": 128},
}

n_envs = 6
n_timesteps = 50_000
```

## Level 1

### DummyVecEnv

```python
env = make_vec_env(
    LevelOne, 
    n_envs = n_envs, 
    vec_env_cls = DummyVecEnv, 
    wrapper_class = ImgObsWrapper,
    seed = SEED
)

model_level_path = Path("models/vec_test_dummy.zip")

if model_level_path.exists():
    model = PPO.load(model_level_path)
    print("Model loaded successfully.")
else:
    print("No saved model found. Training a new model ...")
    model = PPO(
        "CnnPolicy", 
        env, 
        policy_kwargs = policy_kwargs, 
        verbose = 1, 
        device = device, 
        seed = SEED
    )  # perhaps lower n_steps from the default 2048
    model.learn(total_timesteps = (n_timesteps * n_envs), progress_bar = True)
    model.save(model_level_path)
```

### SubprocVecEnv

```python
env = make_vec_env(
    LevelOne, 
    n_envs = n_envs, 
    vec_env_cls = SubprocVecEnv, 
    wrapper_class = ImgObsWrapper,
    seed = SEED
)

model_level_path = Path("models/vec_test_subproc.zip")

if model_level_path.exists():
    model = PPO.load(model_level_path)
    print("Model loaded successfully.")
else:
    print("No saved model found. Training a new model ...")
    model = PPO(
        "CnnPolicy", 
        env, 
        policy_kwargs = policy_kwargs, 
        verbose = 1, 
        device = device, 
        seed = SEED
    )  # perhaps lower n_steps from the default 2048
    model.learn(total_timesteps = (n_timesteps * n_envs), progress_bar = True)
    model.save(model_level_path)
```

### Min egna första

```python
env = LevelOne(render_mode="rgb_array")
env = ImgObsWrapper(env)

model_level_path = Path("models/vec_test_albot.zip")

if model_level_path.exists():
    model = PPO.load(model_level_path)
    print("Model loaded successfully.")
else:
    print("No saved model found. Training a new model ...")
    model = PPO(
        "CnnPolicy", 
        env, 
        policy_kwargs = policy_kwargs, 
        verbose = 1, 
        device = device, 
        seed = SEED
    )  # perhaps lower n_steps from the default 2048
    model.learn(total_timesteps = n_timesteps, progress_bar = True)
    model.save(model_level_path)
```

### Verify

```python
obs = env.reset()
while True:
    action, _ = model.predict(obs)
    obs, rewards, terminated, truncated, info = env.step(action)
    env.render("human")
    if terminated or truncated:
        break
```

```python
env.close()
```
